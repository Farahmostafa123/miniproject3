```python
import os, subprocess

DATA_DIR = '/data'
RAW_FILE = f'{DATA_DIR}/ratings_Electronics.csv'
os.makedirs(DATA_DIR, exist_ok=True)

if not os.path.exists(RAW_FILE):
    print('Downloading...')
    url = 'https://snap.stanford.edu/data/amazon/productGraph/categoryFiles/ratings_Electronics.csv'
    subprocess.run(['wget', '-q', '--show-progress', '-O', RAW_FILE, url], check=True)
    print('Done.')
else:
    print('Already exists.')

result = subprocess.run(['wc', '-l', RAW_FILE], capture_output=True, text=True)
print(f'Total rows: {result.stdout.strip()}')
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.types import *
from pyspark.sql.window import Window

spark = (
    SparkSession.builder
    .appName('AmazonRecommender')
    .master('spark://spark-master:7077')
    .config('spark.executor.memory', '1g')
    .config('spark.driver.memory', '1g')
    .config('spark.sql.shuffle.partitions', '8')
    .getOrCreate()
)
spark.sparkContext.setLogLevel('WARN')
print(f'Spark {spark.version} ready')
```

```python
schema = StructType([
    StructField('user_id_raw', StringType(),  True),
    StructField('item_id_raw', StringType(),  True),
    StructField('rating',      FloatType(),   True),
    StructField('timestamp',   LongType(),    True),
])

raw_df = spark.read.option('header','false').schema(schema).csv(RAW_FILE)
print(f'Loaded: {raw_df.count():,}')

# Drop nulls
clean_df = raw_df.dropna(subset=['user_id_raw','item_id_raw','rating','timestamp'])

# Valid ratings only
clean_df = clean_df.filter((F.col('rating') >= 1.0) & (F.col('rating') <= 5.0))

# Remove duplicates — keep latest
w = Window.partitionBy('user_id_raw','item_id_raw').orderBy(F.col('timestamp').desc())
clean_df = (clean_df.withColumn('rn', F.row_number().over(w))
            .filter(F.col('rn')==1).drop('rn'))

# Keep users/items with at least 5 interactions
u_cnt = clean_df.groupBy('user_id_raw').count().filter(F.col('count')>=5)
i_cnt = clean_df.groupBy('item_id_raw').count().filter(F.col('count')>=5)
clean_df = (clean_df
            .join(u_cnt.select('user_id_raw'), on='user_id_raw', how='inner')
            .join(i_cnt.select('item_id_raw'), on='item_id_raw', how='inner'))

print(f'After cleaning: {clean_df.count():,}')
```

```python
from pyspark.ml.feature import StringIndexer
from pyspark.ml import Pipeline

# String → Integer (ALS needs integer IDs)
user_indexer = StringIndexer(inputCol='user_id_raw', outputCol='user_id', handleInvalid='keep')
item_indexer = StringIndexer(inputCol='item_id_raw', outputCol='item_id', handleInvalid='keep')
pipeline      = Pipeline(stages=[user_indexer, item_indexer])
indexer_model = pipeline.fit(clean_df)
indexed_df    = indexer_model.transform(clean_df)
indexed_df    = (indexed_df
                 .withColumn('user_id', F.col('user_id').cast(IntegerType()))
                 .withColumn('item_id', F.col('item_id').cast(IntegerType())))

# Save maps for streaming
user_map_df = spark.createDataFrame(
    [(i,u) for i,u in enumerate(indexer_model.stages[0].labels)], ['user_id','user_id_raw'])
item_map_df = spark.createDataFrame(
    [(i,it) for i,it in enumerate(indexer_model.stages[1].labels)], ['item_id','item_id_raw'])
user_map_df.write.mode('overwrite').parquet('/data/user_map')
item_map_df.write.mode('overwrite').parquet('/data/item_map')
indexer_model.write().overwrite().save('/data/indexer_pipeline')

# User segmentation
user_stats = (
    indexed_df.groupBy('user_id','user_id_raw')
    .agg(F.count('rating').alias('num_ratings'), F.avg('rating').alias('avg_rating'))
    .withColumn('segment',
        F.when(F.col('num_ratings')>=20, 'power_user')
         .when(F.col('num_ratings')>=10, 'regular_user')
         .otherwise('cold_start'))
)
user_stats.write.mode('overwrite').parquet('/data/user_segments')
indexed_df = indexed_df.join(user_stats.select('user_id','segment'), on='user_id', how='left')

print('=== Segment Distribution ===')
user_stats.groupBy('segment').count().orderBy('count', ascending=False).show()
```

```python
from pyspark.ml.recommendation import ALS
from pyspark.ml.evaluation import RegressionEvaluator

ratings_df = indexed_df.select('user_id','item_id','rating').cache()
train_df, test_df = ratings_df.randomSplit([0.8, 0.2], seed=42)
print(f'Train: {train_df.count():,} | Test: {test_df.count():,}')

als = ALS(
    rank=20, maxIter=15, regParam=0.1,
    userCol='user_id', itemCol='item_id', ratingCol='rating',
    coldStartStrategy='drop', nonnegative=True, seed=42
)
print('Training ALS... (5-10 min)')
als_model = als.fit(train_df)

evaluator  = RegressionEvaluator(metricName='rmse', labelCol='rating', predictionCol='prediction')
predictions = als_model.transform(test_df)
rmse = evaluator.evaluate(predictions)
print(f'\n=== RMSE: {rmse:.4f} ===')

# Tune if RMSE > 1.5
if rmse > 1.5:
    print('Tuning...')
    for rank in [10, 30]:
        for reg in [0.05, 0.15]:
            m = ALS(rank=rank, maxIter=15, regParam=reg,
                    userCol='user_id', itemCol='item_id', ratingCol='rating',
                    coldStartStrategy='drop', nonnegative=True, seed=42).fit(train_df)
            r = evaluator.evaluate(m.transform(test_df))
            print(f'  rank={rank}, reg={reg} → RMSE={r:.4f}')
            if r < rmse:
                rmse = r
                als_model = m
    print(f'Best RMSE: {rmse:.4f}')
else:
    print('RMSE acceptable, no tuning needed')
```

```python
import matplotlib.pyplot as plt

als_model.write().overwrite().save('/data/als_model')
print('Model saved')

global_mean = float(indexed_df.agg(F.avg('rating')).first()[0])
popularity_df = (
    indexed_df.join(item_map_df, on='item_id', how='left')
    .groupBy('item_id','item_id_raw')
    .agg(F.count('rating').alias('num_ratings'), F.avg('rating').alias('avg_rating'))
    .withColumn('bayesian_score',
        (F.col('num_ratings')*F.col('avg_rating') + F.lit(50)*F.lit(global_mean))
        / (F.col('num_ratings') + F.lit(50)))
    .orderBy(F.col('bayesian_score').desc())
    .limit(200)
)
popularity_df.write.mode('overwrite').parquet('/data/popularity_fallback')
print('Popularity fallback saved')

# Plot
rating_dist = indexed_df.groupBy('rating').count().orderBy('rating').toPandas()
seg_dist    = user_stats.groupBy('segment').count().toPandas()

fig, axes = plt.subplots(1, 2, figsize=(12,4))
axes[0].bar(rating_dist['rating'], rating_dist['count'], color='steelblue')
axes[0].set_title(f'Rating Distribution | RMSE={rmse:.4f}', fontweight='bold')
axes[1].pie(seg_dist['count'], labels=seg_dist['segment'],
            autopct='%1.1f%%', colors=['#2196F3','#FF9800','#4CAF50'])
axes[1].set_title('User Segments', fontweight='bold')
plt.tight_layout()
plt.savefig('/data/analysis.png', dpi=150)
plt.show()
print('=== Batch phase complete. Ready for streaming ===')
```

```python
from pyspark.ml.recommendation import ALSModel
from pyspark.sql.functions import window
import json, time
from datetime import datetime

als_model       = ALSModel.load('/data/als_model')
popularity_df   = spark.read.parquet('/data/popularity_fallback').cache()
user_seg_df     = spark.read.parquet('/data/user_segments').select('user_id','segment').cache()
item_map_loaded = spark.read.parquet('/data/item_map').cache()

pop_rows = (popularity_df.orderBy(F.col('bayesian_score').desc())
            .limit(5).select('item_id','item_id_raw','bayesian_score').collect())
pop_fallback = [{'item_id':r['item_id'],'item_id_raw':r['item_id_raw'],
                 'score':float(r['bayesian_score'])} for r in pop_rows]

OUTPUT = '/data/output'
for d in ['recommendations','alerts/items','alerts/users','window_stats']:
    os.makedirs(f'{OUTPUT}/{d}', exist_ok=True)

print('Models loaded. Ready to stream.')
```

```python
event_schema = StructType([
    StructField('user_id',     IntegerType(), True),
    StructField('user_id_raw', StringType(),  True),
    StructField('item_id',     IntegerType(), True),
    StructField('item_id_raw', StringType(),  True),
    StructField('rating',      FloatType(),   True),
    StructField('timestamp',   StringType(),  True),
    StructField('segment',     StringType(),  True),
    StructField('event_type',  StringType(),  True),
])

raw_stream = (
    spark.readStream.format('kafka')
    .option('kafka.bootstrap.servers', 'kafka:9092')
    .option('subscribe', 'ecommerce-interactions')
    .option('startingOffsets', 'latest')
    .option('failOnDataLoss', 'false')
    .option('maxOffsetsPerTrigger', 500)
    .load()
)

parsed = (
    raw_stream
    .select(F.from_json(F.col('value').cast('string'), event_schema).alias('e'),
            F.col('partition').alias('kafka_partition'))
    .select('kafka_partition',
            F.col('e.user_id'),   F.col('e.user_id_raw'),
            F.col('e.item_id'),   F.col('e.item_id_raw'),
            F.col('e.rating'),    F.col('e.segment'),
            F.to_timestamp(F.col('e.timestamp'),
                           "yyyy-MM-dd'T'HH:mm:ss'Z'").alias('event_time'))
    .filter(F.col('user_id').isNotNull() & F.col('item_id').isNotNull() &
            F.col('rating').isNotNull() & F.col('event_time').isNotNull())
    .withWatermark('event_time', '2 minutes')
)

# Window analytics
item_windows = (
    parsed.groupBy(window('event_time','30 seconds','10 seconds'),'item_id','item_id_raw')
    .agg(F.avg('rating').alias('avg_rating'),
         F.count('*').alias('interaction_count'),
         F.stddev('rating').alias('rating_variance'))
    .withColumn('engagement_score',
        F.round((F.col('avg_rating')/F.lit(5.0)) * F.log(F.col('interaction_count')+1), 4))
    .withColumn('window_start', F.col('window.start'))
    .withColumn('window_end',   F.col('window.end'))
    .drop('window')
)

user_windows = (
    parsed.groupBy(window('event_time','30 seconds','10 seconds'),'user_id','segment')
    .agg(F.count('*').alias('interaction_count'), F.avg('rating').alias('avg_rating_given'))
    .withColumn('window_start', F.col('window.start'))
    .withColumn('window_end',   F.col('window.end'))
    .drop('window')
)

# Alerts
item_alerts = (
    item_windows
    .filter((F.col('avg_rating') > 4.5) & (F.col('interaction_count') >= 3))
    .withColumn('alert_type', F.lit('TRENDING_ITEM'))
    .withColumn('alert_message',
        F.concat(F.lit('ALERT: Item '), F.col('item_id'),
                 F.lit(' is trending! avg_rating='), F.round(F.col('avg_rating'),2),
                 F.lit(', interactions='), F.col('interaction_count')))
    .withColumn('alert_time', F.current_timestamp())
)

user_alerts = (
    user_windows
    .filter(F.col('interaction_count') > 8)
    .withColumn('alert_type', F.lit('ACTIVITY_SPIKE'))
    .withColumn('alert_message',
        F.concat(F.lit('ALERT: User '), F.col('user_id'),
                 F.lit(' spike! '), F.col('interaction_count'),
                 F.lit(' interactions in 30s')))
    .withColumn('alert_time', F.current_timestamp())
)

print('Streaming pipeline defined.')
```

```python
def process_batch(batch_df, epoch_id):
    t0 = time.time()
    if batch_df.count() == 0:
        return
    batch_users = batch_df.select('user_id','segment').distinct()
    als_users   = batch_users.filter(F.col('segment').isin(['power_user','regular_user'])).select('user_id')
    cold_users  = batch_users.filter(F.col('segment')=='cold_start').select('user_id')
    recs = []
    if als_users.count() > 0:
        als_recs = als_model.recommendForUserSubset(als_users, 5)
        flat = (als_recs
                .withColumn('rec', F.explode('recommendations'))
                .select('user_id',
                        F.col('rec.item_id').alias('item_id'),
                        F.col('rec.rating').alias('predicted_score'))
                .join(item_map_loaded, on='item_id', how='left'))
        for r in flat.collect():
            recs.append({'user_id':r['user_id'],'item_id':r['item_id'],
                         'item_id_raw':r['item_id_raw'],
                         'score':round(float(r['predicted_score']),4),
                         'strategy':'ALS','epoch':epoch_id})
    for row in cold_users.collect():
        for item in pop_fallback:
            recs.append({'user_id':row['user_id'],'item_id':item['item_id'],
                         'item_id_raw':item['item_id_raw'],'score':item['score'],
                         'strategy':'POPULARITY_FALLBACK','epoch':epoch_id})
    latency_ms = round((time.time()-t0)*1000, 1)
    output = {'epoch':epoch_id,'timestamp':datetime.utcnow().isoformat()+'Z',
              'latency_ms':latency_ms,'batch_size':batch_df.count(),'recommendations':recs}
    with open(f'{OUTPUT}/recommendations/epoch_{epoch_id}.json','w') as f:
        json.dump(output, f)
    print(f'[Epoch {epoch_id}] events={batch_df.count()} | latency={latency_ms}ms')

# Start all 4 queries
q_recs = (parsed.writeStream.foreachBatch(process_batch)
          .option('checkpointLocation','/data/checkpoints/recs')
          .trigger(processingTime='10 seconds').start())

q_stats = (item_windows.writeStream.format('json')
           .option('path',f'{OUTPUT}/window_stats')
           .option('checkpointLocation','/data/checkpoints/stats')
           .outputMode('append').trigger(processingTime='10 seconds').start())

q_ialerts = (item_alerts
             .select('alert_type','alert_message','alert_time',
                     'item_id','avg_rating','interaction_count','window_start','window_end')
             .writeStream.format('json')
             .option('path',f'{OUTPUT}/alerts/items')
             .option('checkpointLocation','/data/checkpoints/ialerts')
             .outputMode('append').trigger(processingTime='10 seconds').start())

q_ualerts = (user_alerts
             .select('alert_type','alert_message','alert_time',
                     'user_id','interaction_count','window_start','window_end')
             .writeStream.format('json')
             .option('path',f'{OUTPUT}/alerts/users')
             .option('checkpointLocation','/data/checkpoints/ualerts')
             .outputMode('append').trigger(processingTime='10 seconds').start())

print('All 4 streaming queries running!')
print('Now run Cell 10 (Kafka producer) to send events...')
```

```python
import json, time, random
from datetime import datetime, timedelta
from kafka import KafkaProducer
import pandas as pd

user_map_pd = pd.read_parquet('/data/user_map')
item_map_pd = pd.read_parquet('/data/item_map')
seg_pd      = pd.read_parquet('/data/user_segments')[['user_id','segment']]
user_map_pd = user_map_pd.merge(seg_pd, on='user_id', how='left').fillna({'segment':'cold_start'})

producer = KafkaProducer(
    bootstrap_servers='kafka:9092',
    value_serializer=lambda v: json.dumps(v).encode('utf-8'),
    key_serializer=lambda k: str(k).encode('utf-8'),
    acks='all', retries=3,
)

users = user_map_pd.sample(frac=1).head(500)
items = item_map_pd.sample(frac=1).head(200)

def make_event(u, item_row, late_sec=0):
    ts = datetime.utcnow() - timedelta(seconds=late_sec)
    return {
        'user_id':     int(u['user_id']),
        'user_id_raw': str(u['user_id_raw']),
        'item_id':     int(item_row['item_id']),
        'item_id_raw': str(item_row['item_id_raw']),
        'rating':      round(random.uniform(1.0, 5.0), 1),
        'timestamp':   ts.isoformat() + 'Z',
        'segment':     str(u['segment']),
        'event_type':  'rating',
    }

sent = 0
print('Sending events to Kafka...')
try:
    for i in range(2000):
        evts = []
        u  = users.sample(1).iloc[0]
        ir = items.sample(1).iloc[0]
        late = random.random() < 0.05
        evts.append(make_event(u, ir, late_sec=random.randint(60,120) if late else 0))

        if i % 200 == 0 and i > 0:
            su = users.sample(1).iloc[0]
            print(f'[{i}] ⚡ Spike: user {su["user_id"]}')
            for _ in range(10):
                evts.append(make_event(su, items.sample(1).iloc[0]))

        if i % 300 == 0 and i > 0:
            hot = items.sample(1).iloc[0]
            print(f'[{i}] 🔥 Trending: item {hot["item_id"]}')
            for _ in range(15):
                e = make_event(users.sample(1).iloc[0], hot)
                e['rating'] = round(random.uniform(4.0, 5.0), 1)
                evts.append(e)

        for e in evts:
            partition = 0 if e['segment'] == 'cold_start' else 1
            producer.send('ecommerce-interactions',
                          key=str(e['user_id']), value=e, partition=partition)
            sent += 1

        if i % 50 == 0:
            producer.flush()
            print(f'[{i}/2000] Sent: {sent}')
        time.sleep(0.1)

except KeyboardInterrupt:
    print('Stopped by user.')
finally:
    producer.flush()
    producer.close()
    print(f'Total sent: {sent}')
```
